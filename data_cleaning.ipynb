```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("mental_health_data.csv")
print("Kernel restarted and libraries loaded successfully!")
```

```python
df.isnull().sum()
```

```python
df['gender'].value_counts()
```

```python
df.describe()
```

```python
df['gender'] = df['gender'].replace({'male':'Male', 'F':'Female'})
df['gender'].value_counts()
```

```python
round_cols = ['sleep_hours','work_hours','mood_score','stress_level','physical_activity_hours']
df[round_cols] = df[round_cols].round(1)
```

```python
df.isnull().sum()
```

```python
num_cols = ['sleep_hours', 'work_hours', 'mood_score']

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
num_cols = ['sleep_hours','work_hours','mood_score','stress_level','physical_activity_hours']

for col in num_cols:
    plt.figure(figsize=(6,3))
    sns.histplot(df[col], kde=True)
    plt.title(col)
    plt.show()
```

```python
sns.countplot(data=df, x='gender')
plt.show()

sns.countplot(data=df, x='depression_risk')
plt.show()
```
